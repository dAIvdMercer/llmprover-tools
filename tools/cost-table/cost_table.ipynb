# Cost Table

**What does your workload cost on each model?** Tell it how many input and output tokens a typical run uses, and it shows the cost of that run on every model in a dated price snapshot, cheapest first. Add how many runs you expect per month and it projects the monthly bill. Under a minute, free, no signup.

**No API key needed.** This tool does not call any AI model. It just multiplies the token counts you give it by each model's published price, so nothing you type leaves your machine.

A free tool by [pysolvr / LLM Prover](https://llmprover.pysolvr.com). It is the take-home twin of our live pricing table at [llmprover.pysolvr.com/api-pricing](https://llmprover.pysolvr.com/api-pricing/).

## 2. Describe your workload

A *token* is a chunk of text (roughly 3 to 4 characters, or about three-quarters of a word). Providers bill per token, split into **input** (what you send: your prompt plus any context) and **output** (what the model writes back).

You do not need exact numbers. A rough estimate of a typical run is fine. As a feel: a short question is ~50 input tokens; a page of pasted context is ~1,000; a paragraph answer is ~300 output. If you already run a workload, your provider's dashboard shows your real per-call usage.

- `INPUT_TOKENS`  -- tokens in a typical run's prompt/input.
- `OUTPUT_TOKENS` -- tokens in a typical run's completion/output.
- `RUNS_PER_MONTH` -- optional. Set a number to also project a monthly bill; leave as `None` for per-run cost only.
- `MODELS` -- optional. Leave as `None` to cost every model in the snapshot, or give a list like `['gpt-4o-mini', 'claude-haiku-5-5']` to cost only those.

To find what models exist and what they cost today: **[llmprover.pysolvr.com/models](https://llmprover.pysolvr.com/models/)** and **[llmprover.pysolvr.com/api-pricing](https://llmprover.pysolvr.com/api-pricing/)**.

```python
INPUT_TOKENS  = 1200
OUTPUT_TOKENS = 400

# Optional: expected runs per month. Set a number (e.g. 50000) to project a monthly bill, or None.
RUNS_PER_MONTH = None

# Optional: only cost these model ids. None = every model in the snapshot.
MODELS = None
```

## 3. See the cost across models

The table is sorted cheapest first. You get the cost of one run on each model (and a projected monthly cost if you set `RUNS_PER_MONTH`). **No recommendation** -- this is cost arithmetic only. The cheapest row is the cheapest row, not a judgment of which model is good enough for your task.

```python
from cost_table_core import cost_table_core, format_result

result = cost_table_core(
    input_tokens=INPUT_TOKENS,
    output_tokens=OUTPUT_TOKENS,
    runs_per_month=RUNS_PER_MONTH,
    models=MODELS,
)
print(format_result(result))
```
